```python
r = int(input("Enter number of rats: "))
unit = int(input("Enter food consumed by each rat: "))
n = int(input("Enter number of houses: "))

arr = list(map(int, input("Enter food in each house: ").split()))

if arr is None:
    print(-1)
else:
    food_needed = r * unit
    food = 0
    houses = 0

    for i in range(n):
        food = food + arr[i]
        houses = houses + 1

        if food >= food_needed:
            break

    if food < food_needed:
        print(0)
    else:
        print(houses)
```

```python
n = int(input("Enter a number: "))

output = 0

while n > 0:
    rem = n % 10
    output = output * 10 + rem
    n = n // 10

print("Reversed number =", output)
```

```python
n = int(input("Enter a number: "))
largest = 0
while n > 0:
    rem = n % 10
    if rem > largest:
        largest = rem
    n = n // 10
print("Largest digit =", largest)
```

```python
n = int(input("Enter a number: "))
original = n
reverse = 0
while n > 0:
    rem = n % 10
    reverse = reverse * 10 + rem
    n = n // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
n1 = int(input("Enter number of elements in A: "))
a = []
for i in range(n1):
    a.append(int(input("Enter element: ")))

n2 = int(input("Enter number of elements in B: "))
b = []
for i in range(n2):
    b.append(int(input("Enter element: ")))

c = a + b
c = list(set(c))
c.sort()

n = len(c)

if n % 2 == 1:
    median = c[n // 2]
else:
    median = (c[n // 2 - 1] + c[n // 2]) / 2

print("Median =", median)
```
